```python
class Pedido:
    def __init__(self, folio, producto, cantidad):
        self.folio = folio
        self.producto = producto
        self.cantidad = cantidad

    def mostrar_datos(self):
        print(f"\n--- Datos del Pedido ---")
        print(f"Folio: {self.folio}")
        print(f"Producto: {self.producto}")
        print(f"Cantidad: {self.cantidad}")
        print("------------------------")

    def leer_datos(self):
        print(f"\n[Sistema]: Procesando el pedido {self.folio}...")

# --- Inicio del Programa Interactivo ---
print("SISTEMA DE REGISTRO DE PEDIDOS")

# Datos para el Pedido 1
print("\nIngresa datos para el Pedido 1:")
f1 = input("Folio: ")
p1 = input("Producto: ")
c1 = input("Cantidad: ")
pedido1 = Pedido(f1, p1, c1)

# Datos para el Pedido 2
print("\nIngresa datos para el Pedido 2:")
f2 = input("Folio: ")
p2 = input("Producto: ")
c2 = input("Cantidad: ")
pedido2 = Pedido(f2, p2, c2)

# Ejecución de acciones (Funciones)
pedido1.leer_datos()
pedido1.mostrar_datos()

pedido2.leer_datos()
pedido2.mostrar_datos()

print("\nPrograma terminado y realizado por Hugo Fabian NL =15")
```
